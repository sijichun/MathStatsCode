**实验目的**：调用 dgp_mle_censor.ado 生成区间删失数据，画出直方图与经验分布函数（注意删失点）。

**说明**：为在 notebook 中直接内联显示图形，本 notebook 的 Stata 代码省略了 `.do` 文件中的 `graph export` 导出命令（`code_in_notes` 中对应的 .do 文件保留导出，用于生成书中的插图）。

首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.11`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.11")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

本段调用外部程序 `dgp_mle_censor.ado`（与 do 文件同目录，Stata 会自动在当前工作目录查找）生成**区间删失数据**：`dgp_mle_censor x, obs(500) mu(3.3) sigma2(1)` 生成500个观测，其对数服从 N(3.3, 1)、且落在 [1000,10000]（对数值 [3,4]）之外的观测被删失到端点上。随后用两幅图观察数据的形状：
- `hist x` 直方图（在两端点上会有堆积的“尖峰”，正是删失的痕迹）；`graph export` 导出；
- `sort x`、`gen p=_n/_N` 构造经验分布，用 `line p x, xline(1000 10000, lpattern(dot))` 画经验分布函数并标出两个删失点位置——曲线在 [1000,10000] 之外是平的，说明这部分没有真实观测。
这两幅图帮助理解：删失数据的信息是“不完整”的，直方图与 CDF 的形状与普通数据明显不同。


```python
%%stata
clear
set seed 8855
dgp_mle_censor x , obs(500) mu(3.3) sigma2(1)
hist x
sort x
gen p=_n/_N
line p x, xline(1000 10000, lpattern(dot))
```
