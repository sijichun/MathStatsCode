**生成对数变量**

生成对数收入后用一个 `twoway` 语句叠加两层图：`(hist log_income, bin(100) density)` 画直方图，`density` 选项把纵轴从频数换算为**密度**（保证与密度曲线可比）；`(kdensity log_income)` 画**核密度估计**——核密度是不依赖分组、以每个观测为中心放置平滑核函数后加总得到的非参数密度估计。两者叠加可以同时看到分组的粗略形状与平滑的密度轮廓；`graph export` 导出。


```python
%%stata
gen log_income=log10(labor_inc)
twoway (hist log_income,bin(100) density) (kdensity log_income)
```
