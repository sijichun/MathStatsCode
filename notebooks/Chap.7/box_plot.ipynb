>
> 运行方式：通过 *PyStata* 在 Jupyter 中调用 Stata，安装与使用说明见 `notebook/PyStata.md`。

**实验目的**：绘制对数收入的箱线图，以及按性别分组的并列箱线图，识别离群值。

**说明**：为在 notebook 中直接内联显示图形，本 notebook 的 Stata 代码省略了 `.do` 文件中的 `graph export` 导出命令（`code_in_notes` 中对应的 .do 文件保留导出，用于生成书中的插图）。

首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.7`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.7")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

`use ../datasets/chfs_ind.dta, clear`：载入中国家庭金融调查（CHFS）个体数据。`clear` 表示载入前先清空内存，`../datasets` 说明数据文件位于上一级目录的 datasets 文件夹中。


```python
%%stata
use ../datasets/chfs_ind.dta, clear
```

**生成对数变量**

`gen log_income=log10(labor_inc)`：把劳动收入取以10为底的对数生成新变量——收入类变量通常右偏严重，取对数后分布更接近对称；`label variable` 给变量加中文标签，画图时自动显示。


```python
%%stata
gen log_income=log10(labor_inc)
label variable log_income "对数收入"
```

**箱线图**

`graph box log_income, noout`：画单变量箱线图。箱体上下缘为上下四分位数（中间的线是中位数），须延伸到 1.5 倍四分位距内的最远观测，`noout` 表示不与离群点（箱须之外的个别观测）；`graph export` 导出 PDF。箱线图可以在不损失信息的前提下紧凑地展示分布的中心、离散度与偏度。


```python
%%stata
graph box log_income, noout
```

**加标签**

先给性别变量 a2003 加值标签（1=男性、2=女性），再画 `graph hbox log_income, over(a2003)`：**横向**分组箱线图，每个性别一个箱体并排展示，便于比较两组收入的分布差异。`label def sex 1 "男性"` 定义标签集，`add` 选项在已有标签集上追加，`label values` 把标签集绑定到变量。


```python
%%stata
label def sex 1 "男性"
label def sex 2 "女性", add
label values a2003 sex
graph hbox log_income, over(a2003)
```
