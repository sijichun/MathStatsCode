**实验目的**：将家庭消费支出等距分组，用频数最高一组的中点作为众数的估计。

本段用**分组频数法**估计连续变量的众数：
- `su total_consump` 得到消费支出的最小值 `r(min)` 与最大值 `r(max)`；
- 设定组数 `n_group=100`，用 `gen group=floor((x-min)/(max-min+0.0001)*100)` 把连续的消费支出等距划入 1~100 组（分母加 0.0001 防止最大值恰好等于 max 时分母为零）；
- `tab group` 输出各组的频数分布，频数最多的组（样本中为第2组）即众数所在组。


```python
%%stata
use ../datasets/chfs_hh.dta, clear
su total_consump
local min_total_consump=r(min)
local max_total_consump=r(max)
local n_group=100
gen group=floor((total_consump-`min_total_consump')/(`max_total_consump'-`min_total_consump'+0.0001)*`n_group')
tab group
```

**样本数最多的组为第2组**

用众数公式给出估计值：**众数 ≈ 组下界 + (众数组序号+0.5)/组数 × 组距**，即用频数最高组的中点作为连续变量众数的估计（对连续变量而言“众数”是密度最高点，需要先分组才能定义）；最后 `su total_consump, de`（`detail` 选项）输出包括中位数、四分位数在内的详细描述统计，便于与中位数等位置度量作对比。


```python
%%stata
di "众数=" `min_total_consump'+(2+0.5)/`n_group'*(`max_total_consump'-`min_total_consump'+0.0001)
su total_consump, de
```
