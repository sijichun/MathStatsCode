**实验目的**：利用逆变换法 x=-ln(U)（U~U(0,1)），从均匀分布随机数生成指数分布随机数。


首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.5`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.5")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

关闭分页、清空内存，创建含100个观测的数据集，并用 `set seed 250211` 设定随机数种子，保证每次运行产生的随机数完全相同（结果可复现）。


```python
%%stata
set more off
clear
set obs 100
set seed 250211
```

**产生随机数**

`gen x = -1*log(runiform())`：这是**逆变换法**（inverse transform method）的最简单例子。若 U~U(0,1)，则 -ln(U) 服从参数为1的指数分布 Exp(1)：因为指数分布的分布函数为 F(x)=1-e^{-x}，其反函数为 F⁻¹(u)=-ln(1-u)，而 1-U 与 U 同分布，故直接用 -ln(U) 即可。Stata 的 `runiform()` 不会取到0，因此取对数不会出错；这个技巧可推广到任意已知分布函数的分布。


```python
%%stata
gen x=-1*log(runiform())
```
