**实验目的**：演示设置随机数种子并产生均匀分布随机数 U(0,1)。


首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.5`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.5")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

**关闭--more--**

`set more off`：关闭 Stata 的分页显示，避免输出较长时被 `--more--` 暂停。


```python
%%stata
set more off
```

**清除工作区内所有数据**

`clear`：清空当前内存中的数据，保证从干净状态开始。


```python
%%stata
clear
```

**设置样本量**

`set obs 100`：把数据集的观测数设为 100（相当于样本量 n=100）。注意此时还没有任何变量，`set obs` 是生成模拟数据前的第一步。


```python
%%stata
set obs 100
```

**设置seed**

`set seed 505`：设定随机数种子。Stata 的随机数是伪随机数，固定种子后每次运行都得到**完全相同**的随机数序列，便于教学演示和结果复现（不设种子时每次运行结果都不同）。


```python
%%stata
set seed 505
```

**产生随机数**

`gen x = runiform()`：生成变量 x，`runiform()` 返回 [0,1) 区间上均匀分布的伪随机数。均匀分布是 Stata 全部随机数生成器的基础——其他分布（正态、t、χ² 等）本质上都是由均匀随机数变换而来，这也是本节的出发点。


```python
%%stata
gen x=runiform()
```
