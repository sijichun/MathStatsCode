**实验目的**：绘制对数消费与对数收入的散点图并叠加线性拟合线。

**说明**：为在 notebook 中直接内联显示图形，本 notebook 的 Stata 代码省略了 `.do` 文件中的 `graph export` 导出命令（`code_in_notes` 中对应的 .do 文件保留导出，用于生成书中的插图）。

首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.7`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.7")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

载入家庭数据后，本段展示**消费与收入的关系**：`gen log_total_consump=log(total_consump)` 与 `gen log_total_income=log(total_income)` 生成对数消费与对数收入（对数-对数模型中的斜率近似为弹性）。`twoway` 叠加两个图层：`(scatter ..., if mod(_n,10)==0)` 只画每10个观测中的1个——数据量太大时散点会糊成一片，抽样绘图是常用技巧；`(lfit ...)` 添加线性拟合线，观察对数消费与对数收入的平均关系（斜率即弹性估计）。`legend(pos(11) ring(0))` 把图例放在绘图区内右上角，`label(1 "对数消费")` 等自定义图例文字。


```python
%%stata
use ../datasets/chfs_hh.dta, clear
gen log_total_consump=log(total_consump)
gen log_total_income=log(total_income)
twoway (scatter log_total_consump log_total_income if mod(_n,10)==0) (lfit log_total_consump log_total_income), legend(pos(11) ring(0) label(1 "对数消费") label(2 "拟合"))
```
