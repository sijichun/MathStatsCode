**实验目的**：用直方图对比劳动收入原值与对数收入（log10）的分布，观察对数变换后分布形态的变化。

**说明**：为在 notebook 中直接内联显示图形，本 notebook 的 Stata 代码省略了 `.do` 文件中的 `graph export` 导出命令（`code_in_notes` 中对应的 .do 文件保留导出，用于生成书中的插图）。

首先启动 Stata 环境，并把工作目录切换到 `code_in_notes/Chap.7`，这样 Stata 中的相对路径（如 `../datasets/...`）与同名 do 文件完全一致，图片也会输出到该目录。


```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.7")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

本段用**对比直方图**展示对数变换的效果：先载入数据；`hist labor_inc, name(income_nolog) bin(100)` 画劳动收入的直方图并存名为 income_nolog 的图对象；`gen log_income=log10(labor_inc)` 生成对数收入并同样画直方图（`name(income_log)`）；`graph combine` 把两张图并排合并；`graph export` 导出。`bin(100)` 指定直方图的组数。对比可发现：原始收入严重右偏、大量观测挤在左侧，取对数后分布明显对称、更接近正态——这正是计量经济学中收入变量常取对数的原因。最后 `su log_income` 给出对数收入的基本统计量。


```python
%%stata
use ../datasets/chfs_ind.dta, clear
hist labor_inc, name(income_nolog) bin(100)
gen log_income = log10(labor_inc)
hist log_income, name(income_log) bin(100)
graph combine income_nolog income_log
su log_income
```
