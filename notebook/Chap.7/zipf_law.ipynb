本段用城市人口数据检验**齐夫定律（Zipf's law）**：
- 载入城市数据、保留2011年；`sort v87` 按人口升序排序，`gen rank=_N-_n+1` 把行号倒过来得到“人口从大到小”的排名（最大城市 rank=1）；
- `gen log_rank=log10(rank)` 与 `gen log_pop=log10(v87*10000)`（数据单位为万人，乘10000换算成人）分别取对数；
- `twoway` 画两个散点层：主体部分（`_n<_N-5`，除最大的5个城市）与最大的5个城市（`_n>=_N-5`）——后者用 `mlabel(City)` 单独标注城市名（`mlabel` 给散点加标签），便于识别偏离直线的头部城市；
- 若齐夫定律成立，人口对数对排名对数的散点应大致落在一条斜率为负的直线上（即 log(人口) 与 log(排名) 呈线性关系）。


```python
%%stata
clear
use ../datasets/citydata.dta
keep if Year==2011
sort v87
gen rank=_N-_n+1
gen log_rank=log10(rank)
// 数据的单位为万人
gen log_pop=log10(v87*10000)
twoway (scatter log_pop log_rank if _n<_N-5) (scatter log_pop log_rank if _n>=_N-5, mlabel(City)), legend(off) xtitle("城市大小排序的对数（从大到小）") ytitle("人口的对数")
```
